```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Updated filename to match the sidebar in image_899e30.jpg
file_path = '/content/vgsales.csv'

# Load the dataset
try:
    df = pd.read_csv(file_path)
    print("Success: vgsales.csv loaded!")

    # --- Regional Sales Correlation ---
    plt.figure(figsize=(10, 8))
    sns.heatmap(df[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].corr(), annot=True, cmap='coolwarm')
    plt.title('Regional Sales Correlation Heatmap')
    plt.show()

    # --- Top 10 Publishers by Global Sales ---
    plt.figure(figsize=(12, 6))
    top_publishers = df.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False).head(10)
    top_publishers.plot(kind='bar', color='teal', edgecolor='black')
    plt.title('Top 10 Publishers by Total Global Sales')
    plt.ylabel('Sales (Millions)')
    plt.xticks(rotation=45)
    plt.show()

except FileNotFoundError:
    print("Error: Still can't find the file. Make sure 'vgsales.csv' is visible in the left sidebar.")
```
